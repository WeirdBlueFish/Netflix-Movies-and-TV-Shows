# Netflix Titles EDA:

```
# imports:

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```
df = pd.read_csv('netflix_titles.csv')
```

### Pre-Proccesing:

```
df.info()
```

```
df
```

```
# Date format:

df['date_added'] = pd.to_datetime(df['date_added'].str.strip())
```

```
df['year_added'] = df['date_added'].dt.year
df['month_added'] = df['date_added'].dt.month
```

```
# Title Added Trend:

content_trend = df.groupby(['year_added', 'type']).size().unstack().fillna(0)
print(content_trend.tail())
```

```
# Movies Durations:

df_movies = df[df['type'] == 'Movie'].copy()
df_movies['duration_min'] = df_movies['duration'].str.replace(' min', '').astype(float)

print(df_movies['duration_min'].describe())
```

```
print(df.isnull().sum())
```

```
df['director'] = df['director'].fillna('Unknown')
df['cast'] = df['cast'].fillna('Unknown')
df['country'] = df['country'].fillna('Unknown')

df.dropna(subset=['rating', 'duration', 'date_added'], inplace=True)
```
