# Netflix Titles EDA:

```
# imports:
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```
df = pd.read_csv('netflix_titles.csv')
```

### Pre-Proccesing:

```
df.info()
```

```
df
```

```
# Date format:

df['date_added'] = pd.to_datetime(df['date_added'].str.strip())
```

```
df['year_added'] = df['date_added'].dt.year
df['month_added'] = df['date_added'].dt.month
```

```
# Title Added Trend:

content_trend = df.groupby(['year_added', 'type']).size().unstack().fillna(0)
print(content_trend.tail())
```

```
# Movies Durations:

df_movies = df[df['type'] == 'Movie'].copy()
df_movies['duration_min'] = df_movies['duration'].str.replace(' min', '').astype(float)

print(df_movies['duration_min'].describe())
```

```
# Handling Missing Values:

print(df.isnull().sum())

df['director'] = df['director'].fillna('Unknown')
df['cast'] = df['cast'].fillna('Unknown')
df['country'] = df['country'].fillna('Unknown')

df.dropna(subset=['rating', 'duration', 'date_added'], inplace=True)
print(df.isnull().sum())
```

```
# Top Countries

top_coutries = df['country'].apply(lambda x: x.split(',')[0]).value_counts()
top_coutries.head()
```

### Visualisation:

```
sns.set_theme(style="whitegrid")
plt.figure(figsize=(17, 7))

plt.plot(
    content_trend.index,
    content_trend['Movie'],
    marker='o',
    color='#E50914',
    label='Movie',
    linewidth=2
)
plt.plot(
    content_trend.index,
    content_trend['TV Show'],
    marker='s',
    color='#221F1F',
    label='TV Show',
    linewidth=2
)

plt.title('Netflix Movies & TV Shows Trend')
plt.xlabel('Year')
plt.ylabel('Number of Movies/TV Shows')
plt.legend()

plt.tight_layout()
plt.show()
```
